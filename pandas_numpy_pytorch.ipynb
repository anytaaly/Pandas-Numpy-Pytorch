# Data Vizualization

What Type of data are you working with?

* **Time-Series** : data that spans across continuius time period)
Area Chart

* **Categorical** : Data that can be split up into groups or categories.Pie Chart - Sort the slices, Keep them under ~ 5 and focus on one.

Heat Map, Bar Chart, 100% stacked

* **Numeric** : Data with quantitative values, etiher discrete or continuous.

Line Chart (The dates must be continuous)
Scatter Plot

* **Hierarchical** : Data that has natural groups and sub-groups.


What do you want to communicate?

* **Comparison** Compares values over time or across categories. it can answer questions like: So are we interested in visualizing whether or not our sales this year were greater than last year or Are we interested in determining if our store in New York City is outperforming our store in Los Angeles.

Line Chart ,
Bar Chart (baseline must start at zero) ,
Area Chart ,
100% stacked

* **Composition** Breaks down the component parts of a whole. So instead of comparing our sales in New York directly to our sales in Los Angeles, we might want to understand how each store contributes to our overall sales.

Pie Chart - Sort the slices, Keep them under ~ 5 and focus on one,
100% stacked

* **distribution** Shows the frequency of values within a series.
We might also be interested in looking at the distribution of our data. So which values in our data are rare?Which values are common? What's the mean What's the median? What's the 25th percentile, etc.?

Histogram (avoid using too many bins) ,


* **Relationship** Shows the correlation between multiple variables.
So as our sales in New York City rise to our sales in Los Angeles see a similar increase, are they for some reason decreasing?

Scatter Plot : Remember that correkation does not imply causation.
